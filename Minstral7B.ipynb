```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

```python
print(torch.cuda.is_available())
print(torch.cuda.get_device_name(0))
```

```python
device = torch.device("cuda")
device
```

```python
model = AutoModelForCausalLM.from_pretrained("mistralai/Mistral-7B-Instruct-v0.2")
model.to(device)
```

```python
tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-Instruct-v0.2")
tokenizer
```

```python
messages = [
    # {"role": "assistant", "content": "Hello! How can I be of assistance?"},
    {"role": "user", "content": input("")}
]
```

```python
encoded_message = tokenizer.apply_chat_template(messages, add_generation_prompt=True, return_tensors="pt")
encoded_message
```

```python
model_inputs = encoded_message.to(device)
model_inputs
```

```python
generated_ids = model.generate(input_ids=model_inputs['input_ids'], attention_mask=model_inputs['attention_mask'], max_new_tokens=150, num_return_sequences=5, do_sample=True)
generated_ids
```

```python
tokenizer.batch_decode(generated_ids)
```
